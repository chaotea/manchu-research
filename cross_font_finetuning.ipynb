# Cross-font CRNN adaptation — completed pilot

This notebook displays saved results; opening/running its report cell does not train. Dataset and training were executed separately using `experiments/cross_font_adaptation.py`. The original epoch 13 checkpoint and earlier final-test result are preserved.

300 new training forms × 9 fonts + 900 source-replay presentations from 300 unique images; 120 separate development forms × 9 fonts. A further 150 forms × 11 fonts remain reserved and unrendered. GuFeng and XingShu have no adaptation-training/development images. This is not proof of font separation from original tyotakuki training. Native RAQM rendering, source/cmap/mapping audit, exact word exclusion rules, hashes and fixed settings are recorded in `data/cross_font_v2/`.

Labels lack independent linguistic verification. Original source data is attributed to tyotakuki; local words.csv is attributed to Wiktionary in the source repository README. Font provenance/licenses remain unresolved. See the planning log for decisions and failures.


```python
from pathlib import Path
import json
ROOT = Path.cwd()
if not (ROOT / "research_runs").exists():
    ROOT = ROOT / "research"
report = json.loads((ROOT / "research_runs/cross_font_v2/results.json").read_text())
print("Selected adaptation epoch:", report["selected_epoch"])
for split in ("mixed_development", "original_development"):
    before = report["frozen"][split]["overall"]
    after = report["adapted"][split]["overall"]
    print(f"{split}: words {before['word_accuracy']:.2%} -> {after['word_accuracy']:.2%}; CER {before['cer']:.3%} -> {after['cer']:.3%}")
print("Per-font adapted development:")
for font, values in report["adapted"]["mixed_development"]["by_font"].items():
    print(font, f"{values['exact_matches']}/{values['n']} words, CER {values['cer']:.2%}")
print(report["test_status"])
print(report["limitations"])
```
