# Cross-style v4 development comparison

Dataset: 1,200 train and 200 disjoint development word forms × nine fonts; 300 cached source images repeated three times per epoch. Development words excluded from original training forms. The reserved final was evaluated after development-based selection; see research_runs/cross_style_v4_final/.

| Run | Selected epoch | Mixed accuracy | Mixed CER | Original accuracy |
| --- | ---: | ---: | ---: | ---: |
| continued | 5 | 71.50% | 7.22% | 97.60% |
| fresh | 16 | 88.78% | 2.38% | 66.40% |

Continued: Adam 1e-4, eight epochs, selected subject to 97.6% original-source accuracy floor. Fresh: Adam 1e-3, sixteen epochs, selected by mixed-development CER without that retention floor. The different schedules and initialization mean this comparison does not isolate initialization alone. Both are single-seed development results. Check learning curves for budget-limited training.

Source mappings and font provenance remain unverified; synthetic results do not establish historical-document performance. Repeated words across fonts are correlated. See individual run directories for histories, checkpoints and predictions.
